### Explore: Create a Genie Agent from data

Most of this lesson happens in the Genie UI. Open it in another tab and follow along.

1. In the left sidebar, click **Genie Agents**.
2. In the top-right of the Genie page, click **+ New**.
3. A *Connect your data* dialog opens. Select **All**, then locate and select:
   - your course catalog (the `learn_get_started_da_YOUR_USERNAME` one) → **`retail`** schema → **`customers`** table.
4. Click **Create** at the bottom.

The agent opens with the chat window on the left and the configuration panel on the right (defaulted to **Configure → About**).

### Explore: Configure the agent: About, Data, Instructions

The right panel under **Configure** has four tabs: **About**, **Sources**, **Instructions**, and **Examples**. Each one corresponds to one of the three pillars in the concept block above. We'll set them all.

#### About: title, warehouse, description, common questions

Click the edit icon next to each section and set the values below, then **Save** to confirm:

| Field | Value |
|---|---|
| Title | **Basic Retail Details** |
| Default warehouse | *Pick the SQL warehouse available in your workspace* |
| Description | *This agent is designed to provide a place to query the details of the customers dataset.* |
| Common questions | *How many customers do we have in CA?* |

Common questions appear as suggestion chips when a user opens the agent, useful for nudging stakeholders toward questions you know Genie can answer well.

<div style="border-left: 4px solid #1976d2; background: #e3f2fd; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#0d47a1; margin-bottom:6px; font-size: 1.05em;">Information: why a clear Description matters</strong>
<div style="color:#333;">

The Description doesn't just help your stakeholders find the right agent. Genie reads it too. Treat it like a one-paragraph charter: <em>what this agent covers, what it doesn't, and the kinds of questions it's designed for</em>. agents with a vague description ("Customer data agent") give vaguer answers than ones with a concrete charter.

</div>
</div>

#### Data: review the tables Genie can use

Click the **Sources** tab inside the Configure panel. You should see `customers` listed. From here you can add or remove tables and (importantly) inspect the column metadata Genie is reading.

<div style="border-left: 4px solid #1976d2; background: #e3f2fd; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#0d47a1; margin-bottom:6px; font-size: 1.05em;">Information: why column comments and keys matter</strong>
<div style="color:#333;">

Genie's grasp of a table is only as good as the metadata around it.
<ul>
  <li><strong>Column comments</strong> tell Genie what a field means in business terms, <em>"Total order amount in USD, including tax"</em> beats a bare <code>total_price</code> column every time.</li>
  <li><strong>Primary and foreign keys</strong> tell Genie how to join tables correctly. Without them, ambiguous joins (the kind you fixed in Lesson 05's lab) become Genie's problem.</li>
  <li><strong>A meaningful schema name</strong> like <code>retail</code> reads as a hint about the domain.</li>
</ul>
In a real production setup, you'd invest a day adding comments and PK/FK declarations to the tables behind a agent, not because it's hard, but because the lift in answer quality is dramatic.

</div>
</div>

#### Instructions: teach Genie your terminology

Click the **Instructions** tab. Add a short, concrete instruction that nudges Genie toward your team's conventions. For this agent, paste:

> *When a question references a "loyalty tier" or "tier", answer using the `loyalty_segment` column. Treat segment 1 as the highest tier and segment 4 as the lowest. If a question is about geography, prefer `state` over `country` unless the user asks otherwise.*

Click **Save**.

<div style="border-left: 4px solid #7b1fa2; background: #f3e5f5; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#4a148c; margin-bottom:6px; font-size: 1.05em;">Notes: instructions are version-controllable, not magical</strong>
<div style="color:#333;">

Instructions are just text. Add them in small, specific increments, one rule per line, with a concrete example where it helps. After a few weeks of use, the <strong>Monitor</strong> tab (next section) will show you which questions Genie is fumbling. Most fumbles point to a missing or under-specified instruction.

</div>
</div>

### Explore: Tour the agent: Chat, Monitor, Benchmark

The buttons across the top center of the agent are the three things you'll do most.

| Button | What it's for |
|---|---|
| **Chat** | Where you and your stakeholders ask questions and read Genie's responses. After publication, "Can View" / "Can Run" users only see this and **History**. |
| **Monitor** | The audit log, every question, who asked it, the answer, and whether the user marked it useful. This is where you spot mis-answered questions to fix with new instructions or example queries. |
| **Benchmark** | A suite of questions you run on a recurring basis to confirm the agent still answers well as the data and metadata evolve. |

The top-right buttons cover the rest: **Start a new chat**, **History** (separate threads), **Configure** (the panel you just used), and **Share**. The kebab `(⋮)` menu offers **Clone**, **Assign Certification**, and **Delete (move to trash)**.

### Explore: Trusted assets: example SQL queries

The third pillar from the concept block is **trusted assets**. There are two flavors; we'll add the more common one, example SQL queries, directly to this agent.

1. Go to the **Configure → Examples** tab.
2. Click **Add** and select **Example query** from the available options (or access it from the kebab menu, depending on your UI).
3. For the question, paste:

   > *How many customers do we have per state?*

4. For the SQL, paste:

   ```sql
   SELECT state, COUNT(*) AS customer_count
   FROM customers
   GROUP BY state
   ORDER BY customer_count DESC;
   ```

5. **Save**.

From now on, when a user asks anything about customer counts by region, Genie has a tested reference for what the right query looks like.

<div style="border-left: 4px solid #7b1fa2; background: #f3e5f5; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#4a148c; margin-bottom:6px; font-size: 1.05em;">Notes: UC SQL functions as the second flavor of trusted asset</strong>
<div style="color:#333;">

The other flavor, outside the scope of this lesson, is a <strong>custom SQL function</strong> registered in Unity Catalog (e.g., a function that returns "loyalty tier" given a row). Once registered, it can be referenced from a Genie Agent and used to answer questions that involve your domain logic. Treat both flavors the same way: small, specific, well-documented assets that teach Genie your team's vocabulary.

</div>
</div>

### Explore: Ask the first question

Time to use what you built. Click **Chat** at the top of the agent, then ask:

> *How many customers do we have in CA?*

Watch Genie respond. It should run a SQL query and return a single number with a short explanation. Try a follow-up:

> *And what's the top loyalty tier among them?*

Notice that Genie respects the *Instructions* you added. It interprets "loyalty tier" as `loyalty_segment` without you re-explaining.

### Explore: Create a Genie Agent from a dashboard

agents don't have to start with raw tables. You can spin one up from any AI/BI Dashboard. Genie picks up the dashboard's datasets automatically.

1. Open the **Retail Dashboard** you built in Lesson 08.
2. Switch to the **Edit Draft** view.
3. Click the kebab `(⋮)` in the top-right and select **Settings and themes**.
4. In the **General** tab, toggle **Enable Genie**.
5. Choose **Auto-generate Genie Agent** when prompted (the alternative is to link an existing agent, useful for reusing one agent across multiple dashboards).
6. Click **Publish** to push the change. Close the Sharing dialog.
7. Switch to **View Published**. You'll see a new **Ask Genie** button at the bottom of the dashboard.
8. Click **Ask Genie**, then ask:

   > *What tables are there and how are they connected? Give me a short summary.*

The agent inherits the dashboard's datasets, so its first answer should describe `sales` and `customers` in terms of the joins your dashboard already uses.

<strong style="display:block; color:#e65100; margin-bottom:6px; font-size: 1.05em;">Warning: auto-generated agents still need configuration</strong>
<div style="color:#333;">

The agent you just auto-generated has no description, no common questions, no instructions, and no example queries, only the dashboard's datasets. It's a useful starting point, not a finished product. Treat the auto-generation as Step 0; revisit <strong>About</strong>, <strong>Instructions</strong>, and <strong>Example SQL queries</strong> before sharing the agent with stakeholders.
