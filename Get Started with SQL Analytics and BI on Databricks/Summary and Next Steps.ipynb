## What You Built in This Course

<div style="background: #F9F7F4; border-radius: 10px; padding: 20px 24px; box-shadow: 0 2px 8px rgba(27,49,57,0.06); border-top: 6px solid #1B5162; margin: 16px 0;">

Over the course of these lessons, you went from a blank workspace to a published, AI-augmented analytics workflow. Specifically, you:

- **Got oriented** in the Databricks Workspace and located data through Catalog Explorer and the three-level namespace.
- **Wrote SQL queries** from a notebook against a Serverless SQL Warehouse, including aggregates, joins, and date-based grouping.
- **Used Genie Code** to generate SQL from natural-language prompts and to diagnose broken queries with the Diagnose error workflow.
- **Worked with Delta Lake**: created tables, modified data with `INSERT` / `UPDATE` / `DELETE`, inspected the change log with `DESCRIBE HISTORY`, time-traveled with `VERSION AS OF`, and recovered from a destructive change with `RESTORE TABLE`.

- **Created Genie Agents** and configured the three pillars that drive answer quality: Unity Catalog metadata, trusted assets (example SQL queries and UC SQL functions), and agent-level instructions.
- **(Bonus)** Completed a capstone that extended a real dashboard with new joined datasets, added a Heatmap and a dual-axis Line Chart, and auto-generated a Genie Agent directly from the dashboard for self-service follow-up questions.

## Keep Using Genie Code Beyond This Course

<div style="background: #F9F7F4; border-radius: 10px; padding: 20px 24px; box-shadow: 0 2px 8px rgba(27,49,57,0.06); border-top: 6px solid #FF5F46; margin: 16px 0;">
<div style="font-size: 14pt; color: #0b2026; line-height: 1.7;">

<strong>Genie Code</strong> lives in every notebook, the SQL editor, and the dashboard authoring surface. Use it to draft SQL from natural language, diagnose error messages, refactor queries, and generate visualizations from prompts.

Two ways to invoke it inside a notebook or SQL cell:

<ul>
  <li>Click the <strong>Genie Code icon</strong> at the top-right of any cell.</li>
  <li>Press <strong>Cmd + I</strong> (macOS) or <strong>Ctrl + I</strong> (Windows / Linux) inside any cell to open the in-cell prompt.</li>
</ul>

In a dashboard, Genie Code surfaces inside each visualization widget. Describe the chart you want, refine the prompt, and let the Configuration Panel handle the polish. The faster you build the habit of describing what you want, the faster the platform can help you.
