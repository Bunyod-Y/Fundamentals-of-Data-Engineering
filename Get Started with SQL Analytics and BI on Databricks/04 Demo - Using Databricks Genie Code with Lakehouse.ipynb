##### EXPAND FOR ADDITIONAL NOTES

<details>

**How to invoke Genie Code in this notebook**

- Click the **Genie Code** icon in the top-right of any code cell, or
- Press **Cmd + I** (macOS) or **Ctrl + I** (Windows / Linux) while a cell is selected.

A prompt box opens directly inside the cell. Type your request in plain English, then click **Generate**. You can **Run suggested**, **Accept**, or **Reject** the result. If the suggestion isn't quite right, click **Edit prompt** to refine it without starting over.

**What Genie Code knows about**

Genie Code is grounded in metadata from Unity Catalog, table names, column names, and column comments. The richer your metadata (good column descriptions, primary/foreign keys, a meaningful schema name), the better its suggestions. You'll see this play out further in Lesson 10 when you build a Genie Agent.

</details>

### Explore: Generate a query from a prompt

Let's start with the *Generate* mode. The next cell is empty. You'll fill it with code that Genie Code drafts for you.

**Step 1:** Click the empty cell below to select it, then either click the **Genie Code** icon in the top-right of the cell or press **Cmd/Ctrl + I**. Paste the following prompt and click **Generate**:

> *Write a SQL query that creates a view called `high_sales` containing every row from the `sales` table where `total_price` is greater than 10000.*

When the suggestion appears, **read it before running**. Make sure the `CREATE VIEW` statement targets your current schema, not a fully qualified path to some other catalog. If it looks right, click **Run suggested**, then **Accept**.

```sql
CREATE OR REPLACE VIEW high_sales AS
SELECT *
FROM sales
WHERE total_price > 10000;
```

<div style="border-left: 4px solid #ff9800; background: #fff3e0; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#e65100; margin-bottom:6px; font-size: 1.05em;">Warning: check the target before you run</strong>
<div style="color:#333;">

If Genie Code generates a fully qualified name like <code>some_catalog.some_schema.high_sales</code>, you may not have write permission there and the query will fail. Edit the prompt to be explicit, for example: <em>"Create the view in my current schema"</em>, or just remove the catalog/schema qualifier from the suggestion before running.

</div>
</div>

**Step 2:** Confirm the view was created. The cell below should return rows where `total_price` is over 10,000.

```sql
SELECT * FROM high_sales LIMIT 10;
```

### Explore: Diagnose broken queries

Now switch to *Diagnose* mode. Each cell below contains a deliberately broken query. **Run it as-is**, watch it fail, then click **Diagnose error** on the error message to see Genie Code's suggested fix. Read the explanation it gives, that's where most of the learning happens, then apply the fix and re-run.

**Error 1: missing keyword.** Run this cell and use Diagnose error on the result.

```sql
SELECT * FROM high_sales total_price > 100
```

**Error 2: missing table reference.** This one looks complete at a glance, but something important is missing.

```sql
SELECT * FROM WHERE total_price > 100;
```

**Error 3: empty SELECT list.** What's the query *trying* to return?

```sql
SELECT FROM customers;
```

**Error 4: aggregate without `GROUP BY`.** This one is sneakier, the syntax is fine but the semantics aren't. Watch how Genie Code's explanation makes the underlying SQL principle clear.

```sql
SELECT product_category, SUM(total_price) AS total_revenue
FROM sales;
```

<div style="color:#333;">

<ul>
  <li><strong>Generate</strong> is fastest when you know <em>what</em> you want but don't remember <em>how</em> to write it (a window function, a date trick, a JOIN pattern).</li>
  <li><strong>Diagnose</strong> shines when an error message is opaque. Even seasoned analysts use it to skip past parsing mental load.</li>
  <li>Both modes work the same way in the standalone <strong>SQL editor</strong> if you'd rather not work in a notebook.</li>
</ul>
