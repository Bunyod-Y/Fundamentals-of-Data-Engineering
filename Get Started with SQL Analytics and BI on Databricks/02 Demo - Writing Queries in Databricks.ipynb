##### EXPAND FOR ADDITIONAL NOTES

<details>

**Common challenges this workflow faces (and how Databricks addresses each one):**

- **Data silos and multiple systems**: needed data is scattered, and teams operate from inconsistent sources of truth. Databricks puts your tables, files, and semantic context in a single governed catalog (Unity Catalog).
- **Data quality**: getting clean, trustworthy data is time-consuming and delays insights. Working from one shared, versioned source means fewer copies and less reconciliation.
- **Governance and compliance**: every platform in the stack has its own access model, making sensitive data hard to manage. Unity Catalog applies a single permission model across tables, files, models, and dashboards.

**Data warehouse vs. data lake, and why the distinction is fading**

Historically, organizations stored *structured* data in a **data warehouse** (optimized for "what happened?" reporting) and *unstructured* data in a **data lake** (the foundation for "what will happen?" predictions). That split forced analysts to copy and reconcile data across systems just to answer cross-cutting questions.

The **lakehouse** approach unifies both. Structured tables and raw files live in one open table format storage layer, governed by Unity Catalog and queryable with ANSI SQL. That's the foundation for everything you do in this lesson, a single place to find, query, and visualize data.

</details>

### Explore: Find your data in the catalog

Every analysis starts with the same question, *where does the data live?* Setup already pointed this notebook at your personal catalog and the `retail` schema, so let's confirm and look at what's there.

**Step 1:** Confirm your current catalog and schema. Setup pointed you at your personal `retail` schema, both should be set already.

```sql
SELECT current_catalog(), current_schema();
```

**Step 2:** List the tables in the current schema. You should see the three tables that setup created for you.

```sql
SHOW TABLES;
```

You should see **`customers`**, **`sales`**, and **`sales_orders`**. You can also see all of this visually in **Catalog Explorer**: click **Catalog** in the left sidebar, expand your catalog, then expand the **`retail`** schema. The same three tables appear, plus a `data` volume containing the CSV files used to build them.

### Explore: Use Cell Focus Mode

When a query gets long, or you need a bigger results panel, use **Focus Mode**. Click the focus icon in the top-right corner of the next cell to expand it, the notebook hides everything else and gives you a half-screen results pane. While in Focus Mode you can still navigate between cells, toggle Genie Code, review cell comments and version history, and resize the output window. Press **Exit Focus Mode** when you're done.

```sql
-- Open this cell in Focus Mode to try it out, then run it.
SELECT
  city,
  COUNT(*) AS customer_count
FROM customers
GROUP BY city
ORDER BY customer_count DESC;
```

### Explore: Visualize a result without changing the SQL

With results on screen, Databricks lets you build a visualization directly from the result set, no separate BI tool, no extra `GROUP BY` in the SQL. You'll build full dashboards in Lesson 08; for now, a single inline chart is enough to see the pattern.

**Step 4:** Run the query below to pull every row from `customers`. We'll visualize the result in the next step.

```sql
SELECT * FROM customers;
```

**Step 5:** In the result panel above:

1. Click the **+** button next to the **Table** tab and select **Visualization**.
2. Leave **Visualization Type** as **Bar**.
3. Set the **X column** to **state**.
4. Click **add column** under **Y column** and select __*__ (asterisk).
5. Leave **Count** selected.
6. Click **Save**.

The visualization is added next to the results, grouped by state, without any aggregation in the SQL itself. Click the title to rename it, or click **Edit Visualization** to refine it further.
