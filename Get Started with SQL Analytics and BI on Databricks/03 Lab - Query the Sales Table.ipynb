### Task 4: Use Cell Focus Mode

Open the next cell in **Focus Mode** (focus icon in the top-right of the cell). Then fill in the query to return the **average sale price per product category**, alongside the count of sales in that category. Use these column aliases:

- `product_category`
- `avg_sale_price` (average of `total_price`)
- `num_sales` (count of rows)

Sort by `avg_sale_price` descending. While in Focus Mode you'll have a wider results panel, useful when you're scanning many rows.

```sql
-- TODO: Average sale price and number of sales per product_category, sorted by avg_sale_price desc
SELECT
  product_category,
  <FILL_IN> AS avg_sale_price,
  <FILL_IN> AS num_sales
FROM sales
GROUP BY product_category
ORDER BY <FILL_IN>;
```

### Task 5: Visualize the top categories

Re-run your Task 3 query (or run the cell below, which is the same query). Then build a bar chart from the result so a stakeholder could see the answer at a glance.

```sql
SELECT
  product_category,
  SUM(total_price) AS total_revenue
FROM sales
GROUP BY product_category
ORDER BY total_revenue DESC;
```

In the result panel above:

1. Click the **+** button next to the **Table** tab and select **Visualization**.
2. Leave **Visualization Type** as **Bar**.
3. Set the **X column** to **product_category**.
4. Click **add column** under **Y column** and select **total_revenue**.
5. Click **Save**.

Click the widget title to rename it to something descriptive, for example, *Total Revenue by Product Category*. You should now see the same data from Task 3 as a sorted bar chart.
