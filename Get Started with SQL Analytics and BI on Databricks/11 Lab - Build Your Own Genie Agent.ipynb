**Where we left off:** In Lesson 10 you built a Genie Agent on the `customers` table and learned what drives answer quality. Unity Catalog metadata, trusted assets, and agent-level instructions. Now your **Sales operations lead** has a different ask:

> *"I want a self-serve agent my team can ask sales questions in. Just transactions for now, product mix, monthly trends, that kind of thing. Set it up and prove it works on three real questions."*

In this practice you'll build a fresh agent from the **`sales`** table (not `customers`), configure it the way Lesson 10 demonstrated, and run a quick three-question benchmark. The `<FILL_IN>` placeholders below describe what *content* to type into the agent's UI fields. There's no SQL to fill in, just well-considered prose.

### Task 1: Create the agent

No FILL_IN, UI steps only:

1. In the left sidebar, click **Genie Agents**, then **+ New** in the top-right.
2. In the *Connect your data* dialog, select **All**, then locate and select your personal catalog → **`retail`** schema → **`sales`** table.
3. Click **Create**.

The agent opens with the chat window on the left and the **Configure → About** panel on the right.

<ul>

#### Benchmark Q3: a *geographic* question (Genie should NOT be able to answer well)

**Your task:** write a question that asks about a geographic dimension, *state*, *country*, or *region*. Replace the placeholder below, then ask it.


The `sales` table doesn't include any geography columns. Watch what Genie does. It should either tell you it can't answer, or (worse) hallucinate a column. **This is the point of the question.** It teaches the most important agent-design principle: *the data you connect determines what Genie can answer*.

<details>
<summary>Hint, and what to do about it</summary>

<em>Which state generates the most revenue?</em>

The fix would be to add the <code>customers</code> table to this agent's <strong>Data</strong> tab. <code>customers</code> has a <code>state</code> column and shares <code>customer_id</code> with <code>sales</code>, so once both are connected Genie can answer questions that span them. ### Task 5: Review the Monitor tab

1. Click **Monitor** at the top of the agent.
2. You should see your three benchmark questions logged with timestamps. Note that Q3 either has a "no data" answer or one that flags missing context.
3. (Optional but recommended) For Q3, click the question and rate the answer down. That signal is what agent owners use to decide which instructions or example queries to add next.

This is the feedback loop in action. After a few weeks of real usage, the **Monitor** tab is the single best source of guidance for what to improve about the agent.
