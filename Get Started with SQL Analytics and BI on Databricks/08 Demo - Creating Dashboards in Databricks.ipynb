### Explore: Create a new dashboard

Most of this lesson happens in the Dashboards UI rather than in this notebook. Open it in another tab and follow along. Keep this notebook open for instructions.

1. In the left sidebar, click **Dashboards**.
2. Click **Create dashboard** in the top-right.
3. At the top of the resulting screen, click the dashboard name (currently *Untitled dashboard*) and rename it to **Retail Dashboard**.

The dashboard opens on the **Untitled page** tab. That's the canvas. Next to it is the **Data** tab, which is where we'll start.

### Explore: Add datasets from the catalog

Click the **Data** tab. On the left, you'll see two icons: **Datasets** (a list, currently empty) and **Catalog** (the standard Catalog browser). Let's add the two source tables this dashboard will draw from.

1. With the **Datasets** icon selected, click **+ Add data**.
2. In the search box, type **`sales`** and pick the `sales` table from your `retail` schema. Click **Confirm**.
3. Repeat for **`customers`**.

Each table is added to your dataset list as a  **Dashboard Local Metric View**. You can use the visual interface to define dimensions, measures, and joins scoped to your dashboard.

<div style="border-left: 4px solid #1976d2; background: #e3f2fd; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#0d47a1; margin-bottom:6px; font-size: 1.05em;">Information: full tables vs. SQL datasets</strong>
<div style="color:#333;">

Adding a whole table is the fastest way to start, but for KPIs and aggregated charts you'll usually want a custom SQL dataset (see the Counter visualization next). Mixing both is fine: keep raw tables for ad-hoc exploration in visualizations, and write SQL datasets for the metrics you want to control precisely.

### Explore: Build a Counter visualization (Sales Goal)

The first visual is a Counter, total sales against a sales goal of $3M. We need a SQL dataset that returns one row with two numbers.

1. On the **Data** tab, click **+ Add SQL dataset**.
2. Paste the following query into the editor:

   ```sql
   SELECT
     SUM(total_price) AS Total_Sales,
     3000000          AS Sales_Goal
   FROM sales;
   ```
3. Select your catalog (**`learn_get_started_da_YOUR_USERNAME`**) and schema (**retail**) 
4. Click **Run** to execute. You should see one row, two columns.
5. Right-click the dataset (or use the kebab `(⋮)` menu) and rename it to **Count Total Sales**.

Now back to the canvas:

6. Click the **Untitled page** tab and select **Add a visualization** from the toolbar at the bottom. Click anywhere on the canvas to drop the widget.
7. In the **Configuration Panel** on the right, set:
   - **Title:** *checked*. Click the placeholder title and rename it to **Sales Goal**.
   - **Visualization:** *Counter*
   - **Value:** *Count Total Sales* > *Total_Sales*
   - **Comparison:** *Count Total Sales* > *Sales_Goal*
8. Click **Total_Sales** in the configuration, choose **Format → Custom**, and set:
   - **Type:** Currency ($)
   - **Abbreviation:** None
9. In the **Style** section, click the **+** next to **Conditional Style** and add: *If Value ≤ Target → Color Red.*

The counter now reads as currency and turns red when total sales fall below the goal, a small, expressive KPI.

### Explore: Build a Combo Chart (three months of sales)

Next, a dual-axis combo chart that shows both *revenue* and *order volume* over a three-month window.

1. On the **Data** tab, click **+ Add SQL dataset** and paste:

   ```sql
   SELECT
     customer_name,
     total_price                       AS Total_Sales,
     date_format(order_date, 'MM')     AS Month,
     product_category
   FROM sales
   WHERE order_date BETWEEN to_date('2019-08-01') AND to_date('2019-10-31');
   ```

2. Select your catalog (**`learn_get_started_da_YOUR_USERNAME`**) and schema (**retail**) .
3. Run it. Rename the dataset to **Three Month Sales**.
4. Back on the canvas, **Add a visualization** and click to drop it.
5. In the **Configuration Panel**, set:
   - **Visualization:** *Combo*
   - **X axis:** *Three Month Sales* > *Month*
   - **Y axis Bar:** *Three Month Sales* > *Total_Sales*, Display Name *Total Sales Value*
   - **Y axis Line:** *Three Month Sales* > *COUNT(\*)*, Display Name *Count of Sales Orders*
6. Open the kebab next to **Y axis** and toggle **Enable dual axis** so each series uses its own scale.
7. Open the kebab next to **Left Y axis (Bar)** → **Format → Custom → Type Currency ($)**.

Notice the pattern: high revenue in months with few orders means the orders that did happen were big-ticket. The chart surfaces that immediately.

### Explore: Generate a chart with Genie Code

This is the most interactive part of the lesson. **Genie Code** in dashboards lets you describe a visualization in plain English; it picks the chart type, columns, and aggregations for you. We'll use it to generate a visualization and then refine it with some manual editing.

#### Prompt: Build the visualization with Genie

1. On the canvas, click **Add a visualization** and drop the widget.
2. At the top of the empty widget there's a prompt input box. Type:

   > *Create a bar chart with product_category on the x-axis and average total_sales on the y-axis*

3. Press **Enter** or click **Submit**. After a short pause, Genie returns a bar chart along with the configuration it chose.

Read the Configuration Panel on the right. It now shows the exact dataset, X column, Y column, and aggregation Genie picked. **Always read this before trusting the chart.** If it picked the wrong dataset (sometimes it lands on the wrong one when both `sales` and *Three Month Sales* exist), switch the dropdown manually.

#### When Genie's draft needs manual finishing

Sometimes Genie gets you 80% of the way and the last 20% is faster in the Configuration Panel. Let's convert the chart we just generated into a *stacked* bar chart with two measures.

1. Click the visualization to focus it; the Configuration Panel populates on the right.
2. If needed, change the **Dataset** to *Three Month Sales* (the column names differ between `sales` and *Three Month Sales*; we'll use the one that already has aliases).
3. Click the kebab next to **Y axis** and toggle **Enable dual axis**.
4. Add a second **Y axis** with *Total_Sales* again.
5. Click that second Y axis label and change its transformation to **MIN**.


You now have a dual-axis bar chart that compares average and minimum sale price per category, a question that's hard to write SQL for, easy to *describe*, and even easier to refine after the AI gets you started.

<div style="border-left: 4px solid #7b1fa2; background: #f3e5f5; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#4a148c; margin-bottom:6px; font-size: 1.05em;">Notes: writing better Genie prompts for visualizations</strong>
<div style="color:#333;">

<ul>
  <li><strong>Name the chart type explicitly</strong> when you have a preference, "bar chart", "line", "scatter", "heatmap". Otherwise Genie picks one based on the data shape, which is usually fine but not always what you want.</li>
  <li><strong>Reference column names directly</strong> (<code>product_category</code>, <code>order_date</code>). Genie reads them from the dataset definition, so the more specific you are, the cleaner the result.</li>
  <li><strong>State the aggregation</strong>, "average", "sum", "count distinct", to avoid surprises.</li>
  <li><strong>Include any sort, limit, or filter</strong> as part of the prompt, easier than fixing it in the Configuration Panel afterward.</li>
  <li><strong>Switch datasets if the suggestion looks wrong</strong>. Genie occasionally binds to a sibling dataset; the fix is one dropdown click.</li>
</ul>

</div>
</div>

### Explore: Build a Scatterplot manually

Not every chart needs Genie. For a quick scatter of revenue over time, manual config is faster.

1. **Add a visualization** and drop it on the canvas.
2. In the Configuration Panel:
   - **Visualization:** *Scatter*
   - **X axis:** *sales* > *order_date*, Scale Type *Continuous*, Transform *DAILY*
   - **Y axis:** *sales* > *total_price*, Scale Type *Continuous*, Transform *SUM*

You'll see daily revenue over the data's full date range, useful for spotting outlier days at a glance.

### Explore: Add a Text Box header

Add a title to the dashboard so viewers know what they're looking at.

1. Click the **Add a text box** icon in the bottom toolbar and drop the widget at the top of your canvas.
2. Type: `# Retail Organization`

The `#` makes it a level-1 heading, text boxes use Markdown. Use this widget for section headers, narrative captions, or links to related resources.

### Explore: Publish and share the dashboard

When you're done designing, publish the dashboard so others can view it.

1. Click **Publish** in the top-right.
2. In the dialog, choose **Share data permission** (default). With this option, viewers query through your data permissions, which enables a shared cache and faster loads. The alternative, *Individual data permission*, uses each viewer's own grants and refreshes more often.
3. Optionally notify people via email and add a release-note message.
4. Click **Publish**. The **Sharing** dialog opens.
5. Search for individuals, groups (like *All workspace users*), or set **Anyone in my account can view**: the last option is what makes account-wide access possible without needing workspace assignment.

Use the context icon next to the dashboard name at the top to flip between **Edit Draft** and **View Published**. Edits to the draft don't appear to viewers until you republish.

<div style="border-left: 4px solid #ff9800; background: #fff3e0; padding: 14px 18px; border-radius: 4px; margin: 16px 0;">
<strong style="display:block; color:#e65100; margin-bottom:6px; font-size: 1.05em;">Warning: viewers see the published version, not your draft</strong>
<div style="color:#333;">

When you tweak a draft, viewers keep seeing the previously published version until you republish. <em>Data</em> refreshes happen automatically on the published dashboard as new rows arrive in the underlying tables, only <em>structural</em> changes (new charts, edited queries, layout) require a republish.

### Explore: Add an interactive Filter

A static dashboard answers the questions you anticipated. A *filtered* dashboard lets viewers slice the data themselves.

1. Switch back to the **Edit Draft** view if you're not already there.
2. Click the **Filter** icon in the bottom toolbar and drop the widget under your text-box header.
3. With the filter selected, configure on the right:
   - **Filter type:** *Single value*
   - **Fields:** add both **sales** > **product_category** and **Three Month Sales** > **product_category**, the filter applies to every visualization built on those datasets.
4. Toggle **Title** on, then double-click the widget title and change it to **Product Category**.
5. Use the dropdown in the filter widget to test it, the visualizations should update.
6. Click **Publish** again to push the filter to viewers.

Field filters are the fastest interactivity to add. **Parameters** (where the value gets substituted into the dataset query at runtime) are more powerful, see the docs below, but beyond the scope of this course.
