### Task 4: Diagnose: fix the ambiguous JOIN

This query was supposed to return each customer alongside their total spend, similar to Task 1 but without the top-10 limit. It runs into a different kind of error. Run it, click **Diagnose error**, read Genie Code's explanation carefully. This one teaches a real SQL principle.

```python
%sql
SELECT customer_id, customer_name, SUM(total_price) AS total_spend
FROM customers
JOIN sales ON customer_id = customer_id
GROUP BY customer_id, customer_name;
```

##### Hint

<details>
  <summary>What's wrong?</summary>

  Both <code>customers</code> and <code>sales</code> have a column called <code>customer_id</code>. When you reference it in the <code>JOIN ... ON</code> clause without qualifying which table, SQL doesn't know which one you mean, the reference is <strong>ambiguous</strong>. The same issue applies to <code>customer_id</code> in <code>SELECT</code> and <code>GROUP BY</code>.
<br/><br/>
The fix is to alias each table and qualify the columns:

<button onclick="copyBlock()">Copy to clipboard</button>

<pre id="copy-block" style="font-family: ui-monospace, SFMono-Regular, Menlo, Monaco, Consolas, 'Liberation Mono', 'Courier New', monospace; border:1px solid #e5e7eb; border-radius:10px; background:#f8fafc; padding:14px 16px; font-size:0.85rem; line-height:1.35; white-space:pre;">
<code>
<!-------------------ADD SOLUTION CODE BELOW------------------->
SELECT
  c.customer_id,
  c.customer_name,
  SUM(s.total_price) AS total_spend
FROM customers c
JOIN sales s ON c.customer_id = s.customer_id
GROUP BY c.customer_id, c.customer_name;
<!-------------------END SOLUTION CODE------------------->
</code></pre>
