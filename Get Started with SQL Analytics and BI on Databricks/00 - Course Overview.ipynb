# Get Started with SQL Analytics and BI on Databricks

## Overview

This course introduces the foundational skills a data analyst needs to do day-to-day SQL analytics and BI work on the **Databricks Data Intelligence Platform**. You'll explore the workspace, query data through Unity Catalog, build dashboards, and stand up a conversational Genie Agent, all from a single, governed environment.

The course follows a hands-on path. After getting oriented in the workspace, you'll work through paired **Demo** and **Lab** notebooks for each topic. Demos walk you through a concept with the instructor or a guided notebook; Labs ask you to apply what you just saw on your own.

The content covers:

- The **Databricks Data Intelligence Platform** and how the workspace, **Unity Catalog**, and notebooks fit together.
- Writing SQL queries against Unity Catalog tables from a notebook backed by a **Serverless SQL Warehouse**.
- Using **Genie Code** to generate SQL from natural-language prompts and to diagnose broken queries with the *Diagnose error* workflow.
- Working with **Delta Lake**: `INSERT` / `UPDATE` / `DELETE`, version history with `DESCRIBE HISTORY`, time travel with `VERSION AS OF`, and recovery with `RESTORE TABLE`.
- Building **AI/BI Dashboards** with KPIs, dual-axis charts, Genie-drafted visualizations, and field filters, then publishing them account-wide.
- Creating **Genie Agents** for conversational analytics, configured with Unity Catalog metadata, instructions, and trusted assets.
- **(Bonus)** A capstone that ties dashboards and Genie back together end-to-end on a real business scenario.

By the end of the course, you'll be comfortable discovering data in Catalog Explorer, writing analytics queries against Unity Catalog managed tables, presenting the results in a published dashboard, and enabling stakeholders to ask their own questions through a Genie Agent.

## Terminal Objectives

By the end of this course, you will be able to:

- **Navigate** the Databricks workspace, Unity Catalog, and notebooks to locate and work with data.
- **Develop** SQL queries that aggregate and join Unity Catalog tables from a notebook connected to a Serverless SQL Warehouse.
- **Use** Genie Code to generate, refine, and debug SQL inside notebooks and dashboards.
- **Apply** Delta Lake capabilities, DML, time travel, and schema management, to safely modify and audit table data.
- **Create** AI/BI Dashboards with multiple visualizations and interactive filters, and publish them with appropriate sharing.
- **Configure** Genie Agents with descriptions, instructions, and example queries so stakeholders get high-quality conversational answers.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **Basic familiarity with cloud data platforms.** You understand what a database, table, and query are at a conceptual level.
- **Working knowledge of SQL.** You can read and write `SELECT` statements with `WHERE`, `GROUP BY`, aggregate functions, and `JOIN`. You're familiar with `INSERT`, `UPDATE`, and `DELETE`.
- **General awareness of analytics concepts** like dashboards and BI is helpful but not required, the course covers the basics.
