# Python data types

In Python we have 3 groups of data types: No value, Single value and Multiple values

```python
name  = "Bunyod" # string
age = 1 # integer
pi = 3.14 # float
c = 1j # complex
is_married = False # boolean
list = ['bunyod', 'yokubov'] # list - mutable
tuple = ('hello', 'mother', 'f*cker') # tuple - immutable
dictionary = { "name": "Bunyod", "status": "single"} # dictionary
set = {"apple", "banana", "cherry"} # set
frozenset = ({"apple", "banana", "cherry"})
Nonetype = None # Nonetype
range_nums = range(6) # range
```
